## 1. Open the authoring environment in the browser

Push this repository to GitHub. On the repository page, choose **Code → Codespaces → Create codespace on your default branch**. The included development-container definition installs Python, JupyterLab, Quarto, and the editor extensions automatically.

In the Codespaces terminal, start the live site:

```console
quarto preview --port 4200 --no-browser
```

Port 4200 is forwarded to a browser preview. Saving a notebook triggers a rebuild, so the page updates as you write.

## 2. Write a new post as a notebook

1. Copy `notebook-template.ipynb` into `posts/` and give it a URL-friendly name such as `regularization-intuition.ipynb`.
2. Edit the first **Raw** cell. This cell holds title, description, date, and categories between `---` lines.
3. Write explanations in Markdown cells and experiments in code cells.
4. Run top to bottom before publishing. Hidden state is the most common source of a notebook that works for its author but fails in the cloud.
5. Change `draft: true` to `draft: false` or remove the draft line.

A minimal metadata cell looks like this:

```yaml
---
title: "Regularization without the mystery"
description: "See how a penalty changes fitted coefficients."
date: "2026-09-10"
categories: [regularization, foundations]
---
```

## 3. Publish to GitHub Pages

Commit the notebook and push `main`:

```console
git add posts/regularization-intuition.ipynb
git commit -m "Add regularization notebook"
git push
```

For a direct deployment, render and push the generated site to a `gh-pages` branch with:

```console
quarto publish gh-pages --no-browser
```

The optional `.github/publish-workflow.yml.example` contains the equivalent GitHub Actions automation. After the GitHub credential has `workflow` permission, copy it to `.github/workflows/publish.yml` to rebuild automatically after a push. In either mode, check **Settings → Pages** and select the `gh-pages` branch if GitHub has not selected it automatically.

The official references are [Quarto's notebook authoring guide](https://quarto.org/docs/get-started/authoring/jupyter.html), [blog guide](https://quarto.org/docs/websites/website-blog.html), and [GitHub Pages publishing guide](https://quarto.org/docs/publishing/github-pages.html).

## Authoring cloud versus publishing cloud

These are deliberately separate:

| Concern | Service | What it does |
|---|---|---|
| Authoring | GitHub Codespaces | Gives you Jupyter and a terminal in the browser |
| Building | Quarto CLI or GitHub Actions | Re-runs notebooks in a controlled environment |
| Hosting | GitHub Pages | Serves static HTML, CSS, search data, and RSS |

This separation is useful. A broken or expensive notebook never runs when a reader opens a page, secrets are not shipped to the browser, and hosting remains simple.

## Before turning the sample into a real publication

- Change **Edukron Notes** in `_quarto.yml` if you reuse this academy under another brand.
- Add `site-url` under `website:` after you know the final URL; this makes RSS links absolute.
- Pin package versions more tightly when experiments must remain reproducible for years.
- Never place API keys in a notebook. Store build-time secrets in GitHub Actions secrets and avoid printing them.
- For long or costly experiments, save vetted outputs and use Quarto's `freeze` feature instead of re-running everything on each push.
- Add a custom domain only after the default Pages URL works.

The sample is intentionally small, but the architecture scales: more posts only add more notebooks.
