Machine learning is best learned as a chain of decisions: define an objective, learn parameters, compare against a baseline, inspect failure modes, and decide whether the model is useful. This course builds that chain from first principles.

**Level:** beginner to intermediate  
**Format:** three executable notebooks plus a capstone  
**Prerequisites:** Python, arrays, averages, and basic algebra  
**Estimated time:** 7–9 focused hours

## What you will be able to do

You will implement gradient descent, diagnose learning-rate behavior, build a fair train/test comparison, interpret MAE and RMSE, inspect residuals, standardize features, and implement K-means clustering without hiding the algorithm behind a library call.

## The modeling loop

```{mermaid}
flowchart LR
    A[Question] --> B[Split]
    B --> C[Baseline]
    C --> D[Train]
    D --> E[Evaluate]
    E --> F[Inspect errors]
    F --> G{Useful and reliable?}
    G -- no --> A
    G -- yes --> H[Communicate or deploy]
```

Optimization is only one box. A perfectly optimized model can still answer the wrong question, leak future information, or fail on the group that matters.

## Capstone: make a model earn deployment

Use a regression or classification dataset and write one notebook containing:

- a deployment-shaped question and prediction unit;
- a split that mirrors the intended future use;
- an honest baseline;
- one simple model and one justified improvement;
- aggregate metrics plus an error slice or residual diagnostic;
- a final recommendation: deploy, collect more data, or stop.

The recommendation matters more than squeezing out one extra decimal place.

## Start here

Begin with **[Gradient descent, visually](../../posts/gradient-descent.html)**, then evaluate a model before moving to unsupervised learning.
