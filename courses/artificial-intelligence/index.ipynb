Artificial intelligence includes more than neural networks. An intelligent system needs a representation of its problem, a procedure for choosing an action, and a way to learn or improve. This course connects classical graph search, learning from gradients, and next-token prediction.

**Level:** intermediate  
**Format:** three executable notebooks plus a capstone  
**Prerequisites:** comfortable Python, NumPy arrays, probability basics  
**Estimated time:** 8–10 focused hours

## What you will be able to do

You will represent a problem as states and actions, implement breadth-first and A* search, train a small neural network by backpropagation, build a character language model, and calculate a minimal scaled dot-product attention step.

## Course sequence

| Module | Notebook | Main outcome |
|---:|---|---|
| 1 | [Search with A*](search-and-astar.html) | Turn a world into a graph and guide search with a heuristic |
| 2 | [A neural network from scratch](neural-network-from-scratch.html) | Understand forward passes, loss, gradients, and backpropagation |

## One field, several kinds of intelligence

```{mermaid}
flowchart TD
    AI[Artificial intelligence] --> S[Search and planning]
    AI --> L[Learning from data]
    AI --> G[Generating sequences]
    S --> A[A* chooses a path]
    L --> N[Neural nets learn representations]
    G --> T[Language models predict tokens]
```

The algorithms differ, but the design questions repeat: what is the state, what counts as a good result, what information is available, and how expensive is the search or learning process?

## Capstone: an explainable intelligent system

Choose one route:

- **Planner:** extend A* with weighted terrain and compare heuristics.
- **Learner:** extend the neural network to a small multi-class dataset.
- **Generator:** train the character model on a new, clearly licensed text corpus.

Your notebook must define the objective, visualize at least one internal signal, test a failure case, and explain one safety or reliability limitation.

## Start here

Begin with **[Search with A*](search-and-astar.html)**. It gives you the clearest possible view of an agent, a goal, and the cost of deciding what to do next.
