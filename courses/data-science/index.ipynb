Data science is the discipline of turning observations into decisions without losing track of uncertainty. This course begins with tabular work in pandas, continues through realistic cleaning and exploratory analysis, and ends with a reproducible mini-project.

**Level:** beginner to intermediate  
**Format:** three executable notebooks plus a capstone  
**Prerequisites:** basic Python expressions, loops, and functions  
**Estimated time:** 6–8 focused hours

## What you will be able to do

By the end, you will be able to load and inspect tabular data, select and reshape it, detect common quality failures, build a repeatable cleaning pipeline, choose appropriate summaries and plots, and communicate findings with explicit caveats.

## How to use each notebook

1. Read the question before running anything.
2. Predict the output of each important code cell.
3. Run the notebook top to bottom.
4. Change one input and explain the new output.
5. Complete the **Try it yourself** prompt before moving on.

> Re-running from a clean kernel is part of the lesson. It proves the analysis does not depend on invisible notebook state.

## Capstone: a decision-ready analysis

Choose a small public dataset and answer one bounded question. Your final notebook should contain:

- a one-paragraph question and decision context;
- a data dictionary for the variables you actually use;
- explicit quality checks before and after cleaning;
- at least one distribution and one relationship plot;
- a short conclusion separating evidence, inference, and limitation;
- a reproducibility cell that prints package versions and the final row count.

A strong capstone is not the one with the most charts. It is the one where every transformation and claim has a reason.
