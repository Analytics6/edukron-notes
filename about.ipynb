This is a **notebook-first learning academy**. Every course map and tutorial page is authored as an `.ipynb` file. The site shell—navigation, theme, search, and deployment—is configuration; the lessons, experiments, and exercises remain portable Jupyter notebooks.

> The rule is simple: if it teaches, explains, demonstrates, or documents the project, it belongs in a notebook.

## How one notebook becomes a cloud course

```{mermaid}
flowchart LR
    A[Write .ipynb in Jupyter] --> B[Run cells]
    B --> C[Push to GitHub]
    C --> D[Quarto renders the academy]
    D --> E[Static HTML + search + RSS]
    E --> F[GitHub Pages]
```

The published website is static, so it is inexpensive, quick to load, and has no notebook server exposed to visitors. Readers get the results; authors keep the executable source.

## Editorial principles

1. **Start with the question.** A post should name the decision or confusion it resolves.
2. **Show a baseline first.** Complexity earns its place only after a simple reference point exists.
3. **Keep experiments reproducible.** Seed random generators, declare dependencies, and avoid hidden state.
4. **Interpret every visual.** A chart is not a conclusion. Explain what changed, why it matters, and what remains uncertain.
5. **End with a next move.** Give the reader something small and concrete to try.

## The small stack

| Layer | Purpose |
|---|---|
| Jupyter | Author and execute every content page |
| Quarto | Convert notebooks into a coherent website |
| GitHub Codespaces | Give authors a zero-install cloud workspace |
| Quarto publishing | Rebuild and publish the site to `gh-pages` |
| GitHub Actions | Optional automatic rebuilds after a push |
| GitHub Pages | Serve the generated static site |

You can later swap the last layer for Netlify, Cloudflare Pages, or another static host without rewriting the notebooks.
